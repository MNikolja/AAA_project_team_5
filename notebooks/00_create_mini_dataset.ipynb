# 00 Create Mini Dataset

This notebook takes a small time window from the raw taxi trip data and saves
it as a mini **CSV** dataset (same CSV layout as the raw file). All downstream
notebooks can use this mini data by setting QUICK_RUN = True in the beginning of the code, which makes
notebook 01 read the mini CSV and produce the corresponding _mini parquet.
The mini CSV is also shipped with the repo, so this notebook only needs to be
run if you want to recreate it from scratch. This would mean the full data of 8Gb has to be downloaded as well. We take one week of data so that
the sample covers at least one full weekly cycle. Otherwise weekday/weekend-
specific code downstream could break.

```python
import pandas as pd

# read in the raw data
data = pd.read_csv("../data/Taxi_Trips_(2024-).csv")

# make the start time readable as a date
data["Trip Start Timestamp"] = pd.to_datetime(data["Trip Start Timestamp"])

# pick one week in october 2024
start = pd.Timestamp("2024-10-01")
end = pd.Timestamp("2024-10-08")
mini = data[(data["Trip Start Timestamp"] >= start) & (data["Trip Start Timestamp"] < end)]

# save as csv
mini.to_csv("../data/Taxi_Trips_(2024-)_mini.csv", index=False)

print("total rows:", len(data))
print("mini rows:", len(mini))
```

The mini dataset is ready. Set `QUICK_RUN = True` in notebooks 01 through
the model notebooks to use it.
