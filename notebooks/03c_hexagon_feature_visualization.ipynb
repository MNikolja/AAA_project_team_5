# Hexagon (H3) Feature Visualization Pipeline

Reads `03_merged_data_with_h3.parquet` (the 10% sample enriched with precomputed H3 cell ids), assigns each trip to an **H3 hexagon** (via `start_h3_r9` / `end_h3_r9`, coarsened to the working resolution with `cell_to_parent`), and visualises trip-count and feature distributions per hexagon. This is the hexagon twin of `03a_community_feature_visualization.ipynb`: same pipeline, same plots, with a regular equal-area grid instead of administrative boundaries.

**Caveat — centroid coordinates:** the trip coordinates are census-tract / community-area **centroids**, not raw GPS points, so hexagon binning lights up only the cells that contain a centroid — most hexagons stay empty, especially towards the city edge. The maps below use resolution `H3_RES = 8` (~0.74 km² per cell), a single constant that can be changed in one place.

Hexagons are drawn over the community-area boundaries (light grey) for orientation. All maps are **matplotlib / geopandas** choropleths drawn side by side: **Pickup (left)** and **Dropoff (right)**.

```python
import pandas as pd
import polars as pl
import numpy as np
import h3
import matplotlib.pyplot as plt
from matplotlib.cm import ScalarMappable
from matplotlib.colors import Normalize
from shapely.geometry import shape
import geopandas
```

## Load data

Load only the columns needed for the hexagon analysis. `trip_seconds` is stored as a duration, so we convert it to total seconds for aggregation. The H3 cells are precomputed at resolution 9; we coarsen them to the working resolution `H3_RES` with `cell_to_parent` and report how many distinct hexagons actually receive trips.

```python
H3_RES = 8   # working resolution; r8 cells are ~0.74 km²

_BASE_COLS = [
    "start_h3_r9", "end_h3_r9",
    "trip_start", "trip_end", "hour", "day_of_week", "is_weekend",
    "BankHoliday", "CorporateHoliday", "CityHoliday", "StateHoliday",
]
_FEATURE_COLS = [
    "fare_usd", "tips_usd", "tolls_usd", "extras_usd", "trip_total_usd",
    "trip_miles", "trip_seconds", "average_speed_mph",
    "temperature_2m", "apparent_temperature", "temperature_2m_max", "temperature_2m_min",
    "precipitation", "rain", "relative_humidity_2m", "snow_depth",
    "wind_speed_10m", "wind_direction_10m", "wind_gusts_10m",
]

df = (
    pl.read_parquet("../data/03_merged_data_sample_10pct_with_h3.parquet", columns=_BASE_COLS + _FEATURE_COLS)
    .with_columns(pl.col("trip_seconds").dt.total_seconds().alias("trip_seconds"))
    .to_pandas()
)

# Coarsen the precomputed r9 cells to the working resolution.
for _side in ("start", "end"):
    df[f"{_side}_h3"] = df[f"{_side}_h3_r9"].map(
        lambda c: h3.cell_to_parent(c, H3_RES), na_action="ignore"
    )

print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
print(f"Distinct hexagons at r{H3_RES}: pickup {df['start_h3'].nunique()}, "
      f"dropoff {df['end_h3'].nunique()}")
df.head(3)
```

## Build the hexagon canvas

Unlike 03a/03b there is no fixed shapefile of spatial units: the canvas is every hexagon that appears at least once as a pickup or dropoff cell. Per-subset maps are merged onto this fixed canvas, so hexagons with no trips *in that subset* are drawn in white, exactly like the empty areas/tracts in 03a/03b. The community-area boundaries are loaded as an orientation backdrop.

```python
communities_gdf = (
    geopandas.read_file("../data/chicago_community_boundaries.json")[["geometry"]]
    .set_crs("EPSG:4326")
)

_all_cells = pd.concat([df["start_h3"], df["end_h3"]]).dropna().unique()
hex_gdf = geopandas.GeoDataFrame(
    {"h3_cell": _all_cells},
    geometry=[shape(h3.cells_to_geo([c])) for c in _all_cells],
    crs="EPSG:4326",
)

print(f"{len(hex_gdf)} hexagons in the canvas (r{H3_RES})")
hex_gdf.head(3)
```

## Aggregation & plotting helpers

Identical to the community-area helpers in 03a, except the join key is the H3 cell id, the polygon borders are thinner (many small hexagons instead of 77 large areas) and the community-area boundaries are drawn underneath each map for orientation.

```python
def _hex_aggregate(df_subset, trip_type, feature=None):
    """Aggregate df_subset by H3 hexagon.
    trip_type is 'pickup' or 'dropoff'.
    Returns [h3_cell, trip_count] or [h3_cell, mean_val, trip_count].
    """
    col = "start_h3" if trip_type == "pickup" else "end_h3"
    sub = df_subset.dropna(subset=[col] + ([feature] if feature else []))
    if feature:
        agg = sub.groupby(col)[feature].agg(["mean", "count"]).reset_index()
        agg.columns = ["h3_cell", "mean_val", "trip_count"]
    else:
        agg = sub[col].value_counts().reset_index()
        agg.columns = ["h3_cell", "trip_count"]
    return agg


def _row_geo(sub_pickup, sub_dropoff, kind, feature):
    """Return [pickup_gdf, dropoff_gdf] for one row, with the value column merged on.

    Hexagons with no trips are left as NaN (not filled with 0) so they can be
    drawn explicitly in white via missing_kwds, rather than appearing as the pale
    low end of the colour gradient.
    """
    gs = []
    for subset, trip_type in ((sub_pickup, "pickup"), (sub_dropoff, "dropoff")):
        agg = _hex_aggregate(subset, trip_type, feature if kind == "mean" else None)
        g = hex_gdf.merge(agg, on="h3_cell", how="left")
        gs.append(g)
    return gs


def _value_range(gdfs, value_col, force_zero_min):
    vals = np.concatenate([g[value_col].to_numpy(dtype="float64") for g in gdfs])
    vals = vals[~np.isnan(vals)]
    if vals.size == 0:
        return 0.0, 1.0
    lo = 0.0 if force_zero_min else float(vals.min())
    hi = float(vals.max())
    return lo, (hi if hi > lo else lo + 1)


# One consistent gradient for every trip-count map; the feature maps keep their own (blue) gradient.
# Hexagons with no trips are always drawn in white.
COUNT_CMAP = "YlOrRd"
FEATURE_CMAP = "PuBuGn"
NO_TRIPS_COLOR = "white"
# Many small hexagons vs 77 community areas: thinner borders so fills stay readable.
EDGE_LW = 0.15


def plot_hex_grid(items, kind="count", cmap=COUNT_CMAP, suptitle="",
                  shared_scale=True, cbar_label=None, row_height=3.8):
    """Plot a stacked grid of Pickup (left) / Dropoff (right) choropleths.

    items: list of (label, sub_pickup, sub_dropoff, feature). For counts pass
           feature=None; sub_pickup/sub_dropoff are usually the same dataframe
           except for hour-of-day, where they are filtered by pickup vs dropoff time.
    kind:  'count' (uses trip_count) or 'mean' (uses mean_val of `feature`).

    Hexagons with no trips are rendered in white (NO_TRIPS_COLOR) with a thin grey
    border; the community-area boundaries are drawn underneath for orientation.
    """
    value_col = "trip_count" if kind == "count" else "mean_val"
    force_zero = kind == "count"

    rows = [(label, _row_geo(sp, sd, kind, feat)) for label, sp, sd, feat in items]

    if shared_scale:
        gvmin, gvmax = _value_range([g for _, gs in rows for g in gs], value_col, force_zero)

    def _label(lbl):
        return cbar_label(lbl) if callable(cbar_label) else cbar_label

    n = len(rows)
    fig, axes = plt.subplots(n, 2, figsize=(11, row_height * n),
                             squeeze=False, constrained_layout=True)

    for i, (label, gs) in enumerate(rows):
        vmin, vmax = (gvmin, gvmax) if shared_scale else _value_range(gs, value_col, force_zero)
        for j, g in enumerate(gs):
            ax = axes[i][j]
            communities_gdf.boundary.plot(ax=ax, color="0.8", linewidth=0.4)
            g.plot(column=value_col, ax=ax, cmap=cmap, vmin=vmin, vmax=vmax,
                   edgecolor="0.6", linewidth=EDGE_LW,
                   missing_kwds={"color": NO_TRIPS_COLOR, "edgecolor": "0.6",
                                 "linewidth": EDGE_LW, "label": "no trips"})
            side = "Pickup" if j == 0 else "Dropoff"
            ax.set_title(f"{side} — {label}", fontsize=10)
            ax.set_axis_off()
        if not shared_scale:
            sm = ScalarMappable(norm=Normalize(vmin, vmax), cmap=cmap)
            sm.set_array([])
            cb = fig.colorbar(sm, ax=list(axes[i]), fraction=0.046, pad=0.02)
            if cbar_label is not None:
                cb.set_label(_label(label))

    if shared_scale:
        sm = ScalarMappable(norm=Normalize(gvmin, gvmax), cmap=cmap)
        sm.set_array([])
        cb = fig.colorbar(sm, ax=axes.ravel().tolist(), fraction=0.025, pad=0.02)
        if cbar_label is not None:
            cb.set_label(_label(None))

    if suptitle:
        fig.suptitle(suptitle, fontsize=14)
    plt.show()
```

## Map: trip count per hexagon

Number of trips per hexagon, **Pickup (left)** vs **Dropoff (right)**, hexagons with **no trips are drawn in white**.

```python
#| label: fig-demand-spatial-hex
#| fig-cap: "Trips per H3 hexagon (resolution 8), pickup (left) vs dropoff (right), shared scale. The community-area poles — O'Hare (far north-west) and the downtown lakefront — resolve into individual cells; because the coordinates are area centroids, only cells containing a centroid receive trips."
plot_hex_grid(
    [("All trips", df, df, None)],
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trip count per hexagon",
    shared_scale=True, cbar_label="Trips",
    row_height=5.0,
)
```

## Maps: mean feature value per hexagon

One row per numeric feature; **Pickup (left)** vs **Dropoff (right)**. Hexagons with no trips are shown in **white**.

```python
#| label: fig-feature-grid-hex
#| fig-cap: "Mean value of all 19 numeric features per hexagon, pickup (left) vs dropoff (right), each feature on its own scale. As at community-area and tract level, weather features are near-uniform across the city (a temporal, not spatial, signal) while trip-economics features vary strongly in space; the centroid coordinates leave most cells white."
SELECTABLE_FEATURES = [
    "apparent_temperature", "average_speed_mph", "extras_usd", "fare_usd",
    "precipitation", "rain", "relative_humidity_2m", "snow_depth",
    "temperature_2m", "temperature_2m_max", "temperature_2m_min",
    "tips_usd", "tolls_usd", "trip_miles", "trip_seconds", "trip_total_usd",
    "wind_direction_10m", "wind_gusts_10m", "wind_speed_10m",
]
SELECTABLE_FEATURES = [f for f in SELECTABLE_FEATURES if f in df.columns]

plot_hex_grid(
    [(feat, df, df, feat) for feat in SELECTABLE_FEATURES],
    kind="mean", cmap=FEATURE_CMAP,
    suptitle="Mean feature value per hexagon",
    shared_scale=False,
    cbar_label=lambda feat: f"Mean {feat}",
)
```

### Summary: the distance-to-core gradient

The full feature grid above is comprehensive but too dense for the main report, and it mixes two kinds of feature: **weather**, which is essentially uniform across the city (it is a temporal, not a spatial, signal), and **trip economics**, which varies strongly in space. This row keeps only three trip-economics features that tell one coherent story: mean fare, mean trip distance and mean speed. They show how trip value and length grow with distance from the centre while speed drops in the congested core.

```python
#| label: fig-feature-gradient-hex
#| fig-cap: "Three trip-economics features per hexagon (pickup aggregation): mean fare, mean trip distance and mean speed. The distance-to-core gradient seen at community-area and tract level persists on the regular grid — fare and distance rise toward O'Hare and the periphery while speed falls in the congested core. Each panel keeps its own scale (units differ); the full 19-feature grid is above."

_CURATED = [("fare_usd", "Mean fare (USD)"),
            ("trip_miles", "Mean trip distance (mi)"),
            ("average_speed_mph", "Mean speed (mph)")]

fig, axes = plt.subplots(1, 3, figsize=(13, 5), constrained_layout=True)
for ax, (feat, title) in zip(axes, _CURATED):
    agg = _hex_aggregate(df, "pickup", feat)
    g = hex_gdf.merge(agg, on="h3_cell", how="left")
    communities_gdf.boundary.plot(ax=ax, color="0.8", linewidth=0.4)
    g.plot(column="mean_val", ax=ax, cmap=FEATURE_CMAP,
           edgecolor="0.6", linewidth=EDGE_LW, legend=True,
           legend_kwds={"shrink": 0.6, "label": title},
           missing_kwds={"color": NO_TRIPS_COLOR, "edgecolor": "0.6",
                         "linewidth": EDGE_LW, "label": "no trips"})
    ax.set_title(title, fontsize=11)
    ax.set_axis_off()

fig.suptitle("Trip economics per hexagon — a distance-to-core gradient", fontsize=13)
plt.show()
```

## Maps: trips per hexagon by day of week

One row per weekday; **Pickup (left)** vs **Dropoff (right)**. All days share a colour scale so volumes are comparable across days. `day_of_week` encodes 1 = Monday … 7 = Sunday.

```python
#| label: fig-dow-demand-hex
#| fig-cap: "Trips per hexagon by day of week, pickup (left) vs dropoff (right), shared scale. The weekly rhythm matches the community-area view: weekdays look alike, Saturday softens the O'Hare (airport/business) pole while downtown holds up, and Sunday recovers the airport component."
_DOW_NAMES = {
    1: "Monday", 2: "Tuesday", 3: "Wednesday", 4: "Thursday",
    5: "Friday", 6: "Saturday", 7: "Sunday",
}

_dow_items = [
    (name, df[df["day_of_week"] == i], df[df["day_of_week"] == i], None)
    for i, name in _DOW_NAMES.items()
]

plot_hex_grid(
    _dow_items,
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trips per hexagon by day of week",
    shared_scale=True, cbar_label="Trips",
)
```

## Maps: trips per hexagon by hour of day

One row per hour (00–23); **Pickup (left)** vs **Dropoff (right)**. Pickup maps are filtered by pickup hour, dropoff maps by dropoff hour. All hours share a colour scale.

```python
_dropoff_hour = df["trip_end"].dt.hour

_hour_items = [
    (f"{h:02d}:00", df[df["hour"] == h], df[_dropoff_hour == h], None)
    for h in range(24)
]

plot_hex_grid(
    _hour_items,
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trips per hexagon by hour of day",
    shared_scale=True, cbar_label="Trips",
    row_height=3.4,
)
```

### Summary: taxi demand by hour of day

The 24-row map above is exhaustive but hard to read at a glance. Here we collapse the spatial dimension and keep the temporal one: total pickups and dropoffs per hour across the whole city. This makes the daily cycle legible in a single plot with a near-empty night, a steep morning ramp and a 17:00 evening peak. The pickup and dropoff curves almost overlap; the tiny gap (pickups slightly ahead while demand rises, dropoffs slightly ahead while it falls) is just trip duration pushing dropoffs into later hours, not a strong directional commuting pattern.

```python
#| label: fig-hourly-demand-hex
#| fig-cap: "Total taxi trips by hour of day (10% sample), split by pickup time vs dropoff time. This collapses the 24-row map above onto its time axis (and, with space collapsed, is identical to the community-area version): demand is minimal overnight, ramps up from about 05:00 and peaks at 17:00. The pickup and dropoff curves nearly coincide — pickups run marginally ahead on the rising limb and dropoffs marginally ahead on the falling evening limb, the expected effect of trip durations shifting dropoffs into slightly later hours rather than a strong directional signal."
from matplotlib.ticker import FuncFormatter

# Compact summary of the 24-row hour-of-day maps above: collapse space, keep time.
_pickup_by_hour = df["hour"].value_counts().reindex(range(24), fill_value=0).sort_index()
_dropoff_by_hour = (df["trip_end"].dt.hour
                    .value_counts().reindex(range(24), fill_value=0).sort_index())

# Okabe-Ito blue / orange: a colourblind-safe categorical pair. Marker shape is a
# second, redundant identity cue so the two series never rely on colour alone.
C_PICKUP, C_DROPOFF, INK = "#0072B2", "#E69F00", "#333333"
hours = list(range(24))

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(hours, _pickup_by_hour.values, color=C_PICKUP, lw=2, marker="o",
        markersize=6, label="Pickup")
ax.plot(hours, _dropoff_by_hour.values, color=C_DROPOFF, lw=2, marker="s",
        markersize=6, label="Dropoff")

# One selective callout: the evening pickup peak (the number that matters most).
_peak_h = int(np.asarray(_pickup_by_hour.values).argmax())
_peak_v = int(np.asarray(_pickup_by_hour.values).max())
ax.annotate(f"peak {_peak_h:02d}:00", (_peak_h, _peak_v), color=C_PICKUP,
            fontsize=8, xytext=(0, 9), textcoords="offset points", ha="center")

ax.set_xlim(0, 23)
ax.set_ylim(bottom=0)
ax.set_xticks(range(0, 24, 2))
ax.set_xlabel("Hour of day", color=INK)
ax.set_ylabel("Trips (10% sample)", color=INK)
ax.set_title("Taxi demand by hour of day — pickup vs dropoff", color=INK)
ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v/1000:.0f}k"))
ax.grid(axis="y", color="0.85", linewidth=0.6)
ax.set_axisbelow(True)
for _spine in ("top", "right"):
    ax.spines[_spine].set_visible(False)
ax.tick_params(colors=INK)
ax.legend(frameon=False, loc="upper left")
plt.tight_layout()
plt.show()
```

## Maps: trips per hexagon by 8-hour block

One row per 8-hour block (00–07, 08–15, 16–23); **Pickup (left)** vs **Dropoff (right)**. Pickup maps use pickup time, dropoff maps use dropoff time. Blocks share a colour scale.

```python
#| label: fig-block-demand-hex
#| fig-cap: "Trips per hexagon by 8-hour block, pickup (left) vs dropoff (right), shared scale. Overnight (00-08) is near-empty; midday (08-16) and especially evening (16-24) light up the O'Hare cells for pickups and the downtown core."
_pickup_block = df["hour"] // 8
_dropoff_block = df["trip_end"].dt.hour // 8
_BLOCK_LABELS = {0: "00:00–07:59", 1: "08:00–15:59", 2: "16:00–23:59"}

_block_items = [
    (label, df[_pickup_block == b], df[_dropoff_block == b], None)
    for b, label in _BLOCK_LABELS.items()
]

plot_hex_grid(
    _block_items,
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trips per hexagon by 8-hour block",
    shared_scale=True, cbar_label="Trips",
    row_height=4.2,
)
```

## Maps: trips per hexagon by month

One row per month present in the data; **Pickup (left)** vs **Dropoff (right)**. Pickup maps are filtered by pickup month (`trip_start`), dropoff maps by dropoff month (`trip_end`). Hexagons with no trips are drawn in white.

```python
#| label: fig-month-demand-hex
#| fig-cap: "Trips per hexagon by month, pickup (left) vs dropoff (right), shared scale. The spatial structure is constant all year; volume varies seasonally, with December the lightest month."
import calendar

_pickup_month = df["trip_start"].dt.month
_dropoff_month = df["trip_end"].dt.month
_months = sorted(_pickup_month.dropna().unique())

_month_items = [
    (calendar.month_name[int(m)], df[_pickup_month == m], df[_dropoff_month == m], None)
    for m in _months
]

plot_hex_grid(
    _month_items,
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trips per hexagon by month",
    shared_scale=True, cbar_label="Trips",
)
```

## Maps: trips per hexagon by holiday type

One row per holiday type; **Pickup (left)** vs **Dropoff (right)**. Each trip carries four 0/1 holiday flags (`BankHoliday`, `CorporateHoliday`, `CityHoliday`, `StateHoliday`); a row keeps only the trips where that flag is set. Because holiday days are a small slice of the data, the holiday types differ greatly in total volume, so each row uses its **own** colour scale. Hexagons with no trips are drawn in white.

```python
#| label: fig-holiday-demand-hex
#| fig-cap: "Trips per hexagon by holiday type, pickup (left) vs dropoff (right); each row uses its own scale (holiday types differ greatly in volume). The demand geography is unchanged on holidays — the airport and downtown poles persist."
_HOLIDAY_TYPES = {
    "BankHoliday": "Bank holiday",
    "CorporateHoliday": "Corporate holiday",
    "CityHoliday": "City holiday",
    "StateHoliday": "State holiday",
}

_holiday_items = [
    (f"{name} ({(df[col] == 1).sum():,} trips)", df[df[col] == 1], df[df[col] == 1], None)
    for col, name in _HOLIDAY_TYPES.items()
]

plot_hex_grid(
    _holiday_items,
    kind="count", cmap=COUNT_CMAP,
    suptitle="Trips per hexagon by holiday type",
    shared_scale=False, cbar_label="Trips",
    row_height=4.2,
)
```
